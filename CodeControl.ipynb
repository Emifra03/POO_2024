```python
# estructuras de control
# toma de decisiones (ejecusion condicional)
# if (nos permite ejecuta o ignorar codigo)
valor = 2
if valor < 5:
  print("es muy poquito")
print ("hola")


# elif
# estructura que depende el if, nos ayuda cuando
#tebnemos multiples condiciones

#funciona pero no es lo mejor
valor = 2

if valor <5:
  print("es muy poquito")
if valor > 5:
  print ("es mucho")

#Funciona y es correcto
  valor = 2

if valor <5:
  print("es muy poquito")
elif valor > 5:
  print ("es mucho")
```

```python
# else
# La estructura else se usa en conjunto con if para poder ejecutar un bloque de codigo
# cuando la condicion es falsa
# ejemplo 1
edad = 17

if edad >= 18:
    print("Eres mayor de edad")
else:
    print("Eres menor de edad")
#ejemplo 2
numero = int(input("Ingresa un número: "))

if numero % 2 == 0:
    print("El número es par")
else:
    print("El número es impar")
```

```python
# bucle for
# el bucle for es usado para repetir un bloque de codigo un numero determinado de veces
# o recorrer elementos (listas, cadenas, etc)

# ejemplo 1
suma = 0

for i in range(1, 6):
    suma += i

print("La suma es:", suma)

#ejemplo 2
frutas = ["manzana", "banana", "naranja"]

for fruta in frutas:
    print("Fruta:", fruta)
```

```python
# bucle while
# el bucle while se puede ejecutar mientras una ejecusion sea verdadera

# ejemplo 1
opcion = ""

while opcion != "salir":
    opcion = input("Escribe 'salir' para terminar: ")
    print("Escribiste:", opcion)
# ejemplo 2
contador = 0

while contador < 5:
    print("Contador:", contador)
    contador += 1
```

```python
# listas en python
# la lista son estructuras que permiten alamcenar varios elementos en una sola variable
# se define en corchetes []
# ejemplo 1
frutas = ["manzana", "banana", "naranja"]
print(frutas)

# ejemplo 2
frutas = ["manzana", "banana", "naranja"]

frutas[1] = "pera"
print(frutas)
```

```python
# diccionarios/mapas en python
# los diccionarios almacenan datos en forma clave: valor.
# se define con llaves {}
# ejemplo 1
persona = {
    "nombre": "Juan",
    "edad": 20,
    "ciudad": "Monterrey"
}

print(persona)

# ejemplo 2
alumnos = [
    {"nombre": "Ana", "edad": 20},
    {"nombre": "Luis", "edad": 22},
    {"nombre": "Carlos", "edad": 19}
]

for alumno in alumnos:
    print(alumno["nombre"], "tiene", alumno["edad"], "años")
```
